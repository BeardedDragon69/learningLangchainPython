```python
from ollama import chat
from pydantic import BaseModel


class Explanation(BaseModel):
    topic: str
    difficulty: str
    summary: str


response = chat(
    model="gpt-oss:20b-cloud",
    messages=[
        {
            "role": "user",
            "content": """
            Explain Python lists to a beginner.

            Return ONLY JSON with exactly these fields:
            topic
            difficulty
            summary
            """
        }
    ],
    format=Explanation.model_json_schema()
)

result = Explanation.model_validate_json(response.message.content)

print(result)
print(result.topic)
print(result.difficulty)
print(result.summary)
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama
from pydantic import BaseModel, Field

class Explanation(BaseModel):
    description: str = Field(description="A concise description of the apples")
    number: int 

# Set format="json" and lower temperature to prevent conversational chat
model = ChatOllama(
    model="gpt-oss:20b-cloud",
    temperature=0,
    format="json",
)

# Use input variables so the prompt can be dynamically reused
prompt = ChatPromptTemplate.from_messages([
    ("system", "Extract the description and count according to the schema. Output JSON only."),
    ("user", "{input_text}"),
])

# Use json_mode explicitly if the model struggles with tool-calling
structured_model = model.with_structured_output(Explanation, method="json_mode")

# Chain prompt and model together
chain = prompt | structured_model

response = chain.invoke({"input_text": "I recently bought 20 red apples from the market"})
print(response)
```

```python
from langchain_core.tools import tool
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel


class Answer(BaseModel):
       number1: int 
       number2: int
       answer: str


@tool
def multiply(a: int, b: int) -> int:
       """Multiply two numbers together"""
       print("in tool")
       return a*b



model = ChatOllama(

       model = "gpt-oss:20b-cloud",
       temperature = 0,
       format= "json"

)

messages = ChatPromptTemplate.from_messages([

       ("system", "You are a helpful assistant. You are provided with the multiply tool. Use it whenever necessary. Output JSON only."),
       ("user", "what is 10 multiplied by 10")

])


modelTools = model.bind_tools([multiply])




response = modelTools.invoke(messages.invoke({}))

print(response)
print(response.tool_calls)

```

```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage, ToolMessage
from langchain_ollama import ChatOllama


@tool
def multiply(a: int, b: int) -> int:
    """Multiply two numbers together."""
    return a * b


model = ChatOllama(
    model="gpt-oss:20b-cloud",
    temperature=0
)

model_with_tools = model.bind_tools([multiply])


userMessage = HumanMessage(
       content = "What is 10 multiplied by 10"
)

response = modelTools.invoke([userMessage])

print("MODEL TOOL CALL: ")
print(response.tool_calls)

toolCall = response.tool_calls[0]

toolResult = multiply.invoke(
    toolCall["args"]
)

print("=TOOL RESULT: ")

print(toolResult)


toolMessage = ToolMessage(

    content=str(toolResult),
    tool_call_id = toolCall["id"]

)

messages = [

    userMessage,
    response,
    toolMessage


]

finalResponse = model_with_tools.invoke(messages)


print("FINAL RESPONSE: ")
print(finalResponse.content)
```

```python
from langchain_core.tools import tool
from langchain_ollama import ChatOllama
from langchain_core.prompts import ChatPromptTemplate
from langchain.agents import create_agent

@tool
def multiply(a:float, b:float) -> float:
       """Multiplies 2 numbers and returns the result"""

       return a*b


@tool
def add(a: float, b: float)-> float:
       """Adds two number and returns the sum"""

       return a+b

@tool
def divide(a: float, b: float)-> float:
       """Returns the quotient after dividing a by b"""
       return a/b




#  define the model 
model = ChatOllama(

       model = "gpt-oss:20b-cloud",
       temperature = 0,

)

calculationAgent = create_agent(

       model = model,
       tools = [multiply,add,divide]

)

messages = ChatPromptTemplate.from_messages([

       ("system", "You are a helpful assistant who solves basic arithemetic based on the tools that you are given and returns the answers"),
       ("user", "what is 10*4/5 + 6")

])


prompt = messages.invoke({})

response = calculationAgent.invoke({

       "messages": prompt.messages

})

print(response)


```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver

model = ChatOllama(
       model = "gpt-oss:20b-cloud",
       temperature=0
)

checkpointer = InMemorySaver()

agent = create_agent(

       model= model,
       tools=[],
       checkpointer=checkpointer
)


# the conversation identity/ the thread id

config = {

       "configurable":{
              "thread_id":"conversation-1"
       }

}

response = agent.invoke(
       {
              "messages":[
                     {
                            "role":"user",
                            "content": "My name is sour"
                     }
                     

              ]
       }, config
)


print(response["messages"][-1].content)

response = agent.invoke(
       {
              "messages":[
                     {
                            "role":"user",
                            "content":"What is my name?"
                     }
              ]
       }, config
)

print(response["messages"][-1].content)
```

```python
from langchain_ollama import ChatOllama
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver


model = ChatOllama(
       model = "gpt-oss:20b-cloud",
       temperature=0
)


checkpointer = InMemorySaver()

agent = create_agent(
       model = model,
       tools =[],
       checkpointer = checkpointer
)

name =input("Enter your name: ")
name = name.lower()



config = {
       "configurable":{
              "Id":name
       }
}

while True:
       userInput = input("You: ")

       if userInput.lower == "exit":
              break

       response = agent.invoke(
              {
                     "messages":[
                            {
                                   "role":"user",
                                   "content":userInput
                            }
                     ]

              },config=config
       )


       print("AI:", response["messages"][-1].content)
```
